```python
#Loading Required packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#importing the dataset
df = pd.read_csv("Apple_analysis_clean.csv", sep=";", decimal=",", parse_dates=["Date"])
df = df.sort_values("Date")

df["Daily_Return"] = df["Close"].pct_change()
df["MA_30"] = df["Close"].rolling(window=30).mean()

##Calculating KPIs
daily_std = df["Daily_Return"].std()
annual_volatility = daily_std * np.sqrt(252)
cumulative_return = df["Close"].iloc[-1] / df["Close"].iloc[0] - 1

print("Average daily return:", df["Daily_Return"].mean())
print("Daily standard deviation:", daily_std)
print("Annualized volatility:", annual_volatility)
print("Cumulative return:", cumulative_return)
print("Best daily return:", df["Daily_Return"].max())
print("Worst daily return:", df["Daily_Return"].min())

#Adding charts:
#CHART 1
plt.figure(figsize=(12, 6))
plt.plot(df["Date"], df["Close"], label="Closing Price")
plt.plot(df["Date"], df["MA_30"], label="30-Day Moving Average")
plt.title("Apple Closing Price and 30-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.tight_layout()
plt.savefig("apple_price_trend.png")
plt.show()

#CHART2
plt.figure(figsize=(12, 4))
plt.plot(df["Date"], df["Daily_Return"])
plt.title("Apple Daily Returns")
plt.xlabel("Date")
plt.ylabel("Daily Return")
plt.tight_layout()
plt.savefig("apple_daily_returns.png")
plt.show()
```
